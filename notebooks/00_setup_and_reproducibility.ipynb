# 00 Setup and Reproducibility

This notebook establishes reproducibility and environment checks for the AMZN/MSFT weekend return decision-support project.

- Repo-safe rule: additive workflow only (no edits to existing files).
- Determinism: explicit seeds for Python, NumPy, PyTorch, and RL.
- Dependency checks: verify required libraries before running downstream notebooks.


```python
from pathlib import Path
import os
import random
import importlib

import numpy as np

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)

try:
    import torch
    torch.manual_seed(SEED)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(SEED)
    TORCH_OK = True
except Exception:
    TORCH_OK = False

print({"seed": SEED, "torch_available": TORCH_OK})
```

```python
PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()
OUT_DIR = PROJECT_ROOT / "research_outputs" / "weekend_rl_xgb"
DATA_CACHE_DIR = OUT_DIR / "data_cache"
FIG_DIR = OUT_DIR / "figures"
TABLE_DIR = OUT_DIR / "tables"
MODEL_DIR = OUT_DIR / "models"
for p in [OUT_DIR, DATA_CACHE_DIR, FIG_DIR, TABLE_DIR, MODEL_DIR]:
    p.mkdir(parents=True, exist_ok=True)

print("project_root:", PROJECT_ROOT)
print("output_dir:", OUT_DIR)
```
